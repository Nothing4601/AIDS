```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
n = 500000
elephants = np.random.normal(loc=3350 , scale=500, size=n)
humans = np.random.normal(loc=65, scale=5, size=n)
```

```python
a,bin_human,c =plt.hist(humans,bins=100)
```

```python
d,bin_elephants,e = plt.hist(elephants,bins=100)
```

```python
population = [*elephants ,*humans]
```

```python
len(population)
```

```python
_ = plt.hist(population,bins= [*bin_human,*bin_elephants])
```

```python
def sample_mean_of_data(i,m=10000):  # i is observation in one sample, m is no. of sample
    sample_mean=[]
    for k in range(m):
        sample_mean.append( np.mean(random.sample(population, i)) ) # calculate mean of random smaple which has i observation
    return sample_mean
```

```python
q=sample_mean_of_data(100)
len(q)
```

```python
fig,ax=plt.subplots(3,3,figsize=(14,11))
fig.subplots_adjust(hspace=0.5, wspace=1)
no_of_sample=[[5,9,14],
              [19,25,35],
              [50,70,100]]
for x in range(3):
    for y in range(3):
        ax[x][y].hist(sample_mean_of_data(no_of_sample[x][y]),color='#ADD8E6',bins=100)
        ax[x][y].set_xlabel('avg. weights',color='#8B0000',fontsize=10)
        ax[x][y].set_ylabel('frequency',color='#8B0000',fontsize=10)
        ax[x][y].set_title(f'Distribution of sample mean(n={no_of_sample[x][y]})',color='#8B0000',fontsize=12)
            
```

#### in above we can see that as value of n increase distribution became more normla and it's std is decreases this is CLT!

#### 

```python
def variance_of_sample_mean(i,m=10000):
     std= np.std(sample_mean_of_data(i))
     return std**2
```

```python
n=range(5,100,5)

var_of_mean=[]
formula_var=[]
population_var = (np.std(population))**2
for i in n:
    var_of_mean.append(variance_of_sample_mean(i,m=10))
    formula_var.append(population_var/i)

print('actual var :',var_of_mean)
print('\nformula var :',formula_var)
```

```python
plt.plot(n,var_of_mean,label='actual var')
plt.plot(n,formula_var,label='using formula var')
plt.legend()
plt.title('variance of the averages with the sample size n')
plt.xlabel('no of observation in one sample')
plt.ylabel('var of means')
```
